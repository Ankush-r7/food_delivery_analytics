```python
affected_riders = df[df['lifetime_order_count'].isna()]['rider_id'].unique()

df[df['rider_id'].isin(affected_riders)] \
    .groupby('rider_id')['lifetime_order_count'] \
    .apply(lambda x: pd.Series({
        'Total Orders': len(x),
        'Missing': x.isna().sum(),
        'Present': x.notna().sum()
    }))
```

```python
delivered_missing = df[
    (df['lifetime_order_count'].isna()) &
    (df['cancelled'] == 0)
]

delivered_missing
```

lifetime_order_count is missing for 53 orders belonging to 37 unique riders. For each of these riders, the value is consistently missing across all of their orders, indicating that the missingness is rider-specific rather than order-specific. Although 51 of the affected orders are cancelled, two delivered orders from different riders also have missing values, suggesting that cancellation is not the underlying cause. The missing values most likely result from unavailable rider profile information in the backend.
