```python
# First we will change the datatype of dates and times
```

```python
dates = ['order_time', 'order_date', 'allot_time', 'accept_time', 'pickup_time', 'delivered_time', 'cancelled_time']
```

```python
for col in dates:
    df_cleaned[col] = pd.to_datetime(df_cleaned[col], errors='coerce')
```

```python
df_cleaned['reassigned_order'].fillna(0, inplace = True)
```

```python
df_cleaned.isna().sum()
```

```python
df_cleaned.info()
```

```python
mask = df_cleaned['reassigned_order'].fillna(0) == 0

df_cleaned.loc[mask, 'reassignment_reason'] = (
    df_cleaned.loc[mask, 'reassignment_reason'].fillna('Not Reassigned')
)

df_cleaned.loc[mask, 'reassignment_method'] = (
    df_cleaned.loc[mask, 'reassignment_method'].fillna('Not Reassigned')
)
```

```python
df_cleaned.isna().sum()
```

```python
# Now we will create some new columns
```

```python
df_cleaned.info()
```

```python
# Assignment Delay
df_cleaned["assignment_delay_min"] = (
    df_cleaned["allot_time"] - df_cleaned["order_time"]
).dt.total_seconds() / 60

# Rider Acceptance Time
df_cleaned["acceptance_time_min"] = (
    df_cleaned["accept_time"] - df_cleaned["allot_time"]
).dt.total_seconds() / 60

# Restaurant Preparation Time
df_cleaned["preparation_time_min"] = (
    df_cleaned["pickup_time"] - df_cleaned["accept_time"]
).dt.total_seconds() / 60

# Delivery Time
df_cleaned["delivery_time_min"] = (
    df_cleaned["delivered_time"] - df_cleaned["pickup_time"]
).dt.total_seconds() / 60

# Total Order Fulfillment Time
df_cleaned["total_fulfillment_time_min"] = (
    df_cleaned["delivered_time"] - df_cleaned["order_time"]
).dt.total_seconds() / 60

# Cancellation Delay
df_cleaned["cancellation_delay_min"] = (
    df_cleaned["cancelled_time"] - df_cleaned["order_time"]
).dt.total_seconds() / 60

# Total Distance
df_cleaned["total_distance"] = (
    df_cleaned["first_mile_distance"] +
    df_cleaned["last_mile_distance"]
)

# Order Hour
df_cleaned["order_hour"] = df_cleaned["order_time"].dt.hour

# Day of Week
df_cleaned["day_of_week"] = df_cleaned["order_time"].dt.day_name()

# Weekend Flag
df_cleaned["is_weekend"] = (
    df_cleaned["order_time"].dt.dayofweek >= 5
)
```
